```python
#Câu2
from random import randrange

while True:
    # 1. Khởi tạo game mới
    somay = randrange(1, 101)  # Lấy số ngẫu nhiên từ 1 đến 100
    solandoan = 0
    win = False
    print("\n--- BẮT ĐẦU LƯỢT CHƠI MỚI ---")
    print("(Bạn có 7 lượt đoán)")

    # 2. Vòng lặp cho 7 lần đoán
    while solandoan < 7:
        solandoan += 1
        
        # Thêm try...except để tránh lỗi nếu người dùng nhập chữ
        try:
            songuoi = int(input(f"\n[Lượt {solandoan}/7] Mời bạn đoán số (1-100): "))
        except ValueError:
            print("Lỗi: Bạn phải nhập một SỐ. Bạn đã mất 1 lượt đoán.")
            continue  # Quay lại đầu vòng lặp 'while solandoan < 7'

        # 3. So sánh kết quả
        if somay == songuoi:
            print(f"🎉 CHÚC MỪNG! Bạn đã đoán đúng, số máy là {somay}")
            win = True
            break  # Thoát khỏi vòng lặp đoán (while solandoan < 7)
        elif somay > songuoi:
            print("Bạn đoán sai. Số của máy LỚN HƠN số bạn đoán.")
        else:  # somay < songuoi
            print("Bạn đoán sai. Số của máy NHỎ HƠN số bạn đoán.")

    # 4. Kiểm tra kết quả khi kết thúc 7 lượt
    # Dùng 'if not win:' (nghĩa là win == False)
    if not win:
        print("\nGAME OVER! Bạn đã hết 7 lượt đoán.")
        print(f"Số máy cần tìm là: {somay}")

    # 5. Hỏi chơi lại
    # Dùng .lower() để chấp nhận cả 'k' và 'K'
    hoi = input("\nBạn có muốn chơi tiếp không? (Nhập 'k' để thoát): ")
    if hoi.lower() == "k":
        break  # Thoát khỏi vòng lặp chính (while True)

print("\nCám ơn bạn đã chơi Game! Hẹn gặp lại.")
```

```python
#Câu3
def BMI(height, weight):
    return weight / (height ** 2)

def PhanLoai(bmi):
    if bmi < 18.5:
        return "Gầy"
    elif bmi <= 24.9:
        return "Bình thường"
    elif bmi <= 29.9:
        return "Hơi Béo"
    elif bmi <= 34.9:
        return "Béo Phì Cấp Độ 1"
    elif bmi <= 39.9:
        return "Béo Phì Cấp Độ 2"
    else:
        return "Béo Phì Cấp Độ 3"

def NguyCoBenh(bmi):
    if bmi < 18.5:
        return "Thấp"
    elif bmi <= 24.9:
        return "Trung Bình"
    elif bmi <= 29.9:
        return "Cao"
    elif bmi <= 34.9:
        return "Cao"
    elif bmi <= 39.9:
        return "Rất Cao"
    else:
        return "Nguy Hiểm"

print("Nhập vào chiều cao (m): ")
height = float(input())

print("Nhập vào cân nặng (kg): ")
weight = float(input())

bmi = BMI(height, weight)

print("BMI của bạn =", round(bmi, 2))
print("Phân loại =", PhanLoai(bmi))
print("Nguy cơ bệnh =", NguyCoBenh(bmi))
```

```python
#Câu4
def ROI(dt, cp):
    return (dt - cp) / cp

def GoiYDauTu(roi):
    if roi >= 0.75:
        return "Nên đầu tư"
    else:
        return "Không nên đầu tư"

print("Chương trình tính ROI")
dt = int(input("Nhập Doanh Thu: "))
cp = int(input("Nhập Chi Phí: "))

roi = ROI(dt, cp)
print("Tỉ lệ ROI =", round(roi, 2))
print("==>", GoiYDauTu(roi))
```

```python
#câu7
import math 
def distance(xa, ya, xb, yb):
    return math.sqrt((xb - xa)**2 + (yb - ya)**2)
def main():
    try:
        xa = float(input("Nhập x_A: ").strip())
        ya = float(input("Nhập y_A: ").strip())
        xb = float(input("Nhập x_B: ").strip())
        yb = float(input("Nhập y_B: ").strip())
    except ValueError:
        print("Vui lòng nhập số hợp lệ.")
        return
    d = distance(xa, ya, xb, yb)
    print(f"Dộ dài doạn AB = {d:.4f}")
if __name__ == "__main__":
    main()
```

```python
#câu8
import math
def log_base(a, x):
    #assume a>0, a!=1, x>0
    return math.log(x) / math.log(a)
def main():
    try:
        x = float(input("Nhập x (x>0): ").strip())
        a = float(input("Nhập cơ số a (a>0 và a!=1): ").strip())
    except ValueError:
        print("Vui lòng nhập số hợp lệ.")
        return
    if x<=0:
        print("Lỗi: x phải > 0.")
        return
    if a<=0 or a==1:
        print("Lỗi: a phải > 0 và khác 1.")
        return
    result = log_base(a, x)
    print(f"log_{a}({x}) = {result:.6f}")
if __name__ == "__main__":
    main()
```

```python
#câu9
import math
def S_of_n(n):
    #trả về s(n)=sqrt(2+sqrt(2+..)) với n dấu căn
    if n<= 0:
        raise ValueError("n phải là số nguyên dương (>=1).")
    s = 0.0
    for _ in range(n):
        s = math.sqrt(2+s)
    return s
def main():
    try:
        n = int(input("Nhập n(số dấu căn): ").strip())
    except ValueError:
        print("Vuilonfg nhập số nguyên dương.")
        return
    if n <= 0:
        print("n phải >=1.")
        return
    value = S_of_n(n)
    print(f"S({n})={value:.10f}")
if __name__ == "__main__":
    main()
```

```python
#câu10
import time, os, platform

def clear_screen():
    if platform.system().lower().startswith("win"):
        os.system("cls")
    else:
        os.system("clear")

# 4 hình được mô tả bằng list các dòng (dùng '.' để tương ứng với chấm trong ảnh)
shape1 = [
"      .        ",
"      . .       ",
"      . . .      ",
". . . . . . .     ",
". . .      ",
". .       ",
".        "
]

shape2 = [
"      .        ",
"      . .       ",
"      .   .      ",
". . . . . . .     ",
".   .      ",
". .       ",
".        "
]

shape3 = [
"       . . . .    ",
"       . . .     ",
"       . .      ",
"       .       ",
"     . .      ",
"   . . .     ",
" . . . .    ",
]

shape4 = [
"       . . . .    ",
"       .   .     ",
"       . .      ",
"       .       ",
"     . .      ",
"   .   .     ",
" . . . .    ",
]

shapes = [shape1, shape2, shape3, shape4]

def print_shape(shape):
    for line in shape:
        print(line)

def main(delay=5):
    for i, shape in enumerate(shapes, start=1):
        clear_screen()
        print(f"Hình {i}:\n")
        print_shape(shape)
        time.sleep(delay)
    clear_screen()
    print("Hoàn thành - đã hiển thị 4 hình.")

if __name__ == "__main__":
    main()
```

```python
#câu13
def sum_proper_divisors(n: int) -> int:
    #tính tổng các ước dướng của n nhưng không tính chính n.
    if n<= 1:
        return 0
    s = 1
    import math
    root = int (math.isqrt(n))
    for d in range(2,root+1):
        if n%d == 0:
            s += d
            other = n//d
            if other !=d:
                s +=other
    return s
#Hàm kiểm tra
#kiểm tra số hoàn thiện
def is_perfect(n: int) -> bool:
    return n>0 and sum_proper_divisors(n) == n
#kiểm tra số thịnh vượng
def is_abundant(n: int)->bool:
    return n>0 and sum_proper_divisors(n)>n
#hàm chạy thử
n = int(input("Nhập số nguyên dương n:"))
tong_uoc = sum_proper_divisors(n)
print(f"Tổng các ước (không kể {n})là: ", tong_uoc)

if is_perfect(n):
    print(f"-> {n} Là số hoàn thiện (Perfect number).")
elif is_abundant(n):
    print(f"-> {n} Là số thịnh vượng (Abundant number).")
else:
    print(f"-> {n} Không phải hai loại trên")
```
